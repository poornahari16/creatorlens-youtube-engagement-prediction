### Distribution Analysis

```python
import matplotlib.pyplot as plt

distribution_cols = [
    "view_count",
    "like_count",
    "comment_count",
    "subscriber_count",
    "video_count"
]

for col in distribution_cols:
    plt.figure(figsize=(8, 5))
    plt.hist(df[col].dropna(), bins=50)
    plt.title(f"Distribution of {col}")
    plt.xlabel(col)
    plt.ylabel("Number of Videos")
    plt.tight_layout()
    plt.show()
```

Outlier Analysis

```python
for col in numeric_cols:
    plt.figure(figsize=(8, 5))
    plt.boxplot(df[col].dropna())
    plt.title(f"Box Plot of {col}")
    plt.ylabel(col)
    plt.tight_layout()
    plt.show()
```

```python
# IQR-based outlier counts

outlier_counts = {}

for col in numeric_cols:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outlier_counts[col] = (
        (df[col] < lower_bound) |
        (df[col] > upper_bound)
    ).sum()

print("Potential outlier counts:")
print(pd.Series(outlier_counts))
```

### Correlation Analysis

```python
# Correlation matrix

correlation_cols = [
    "view_count",
    "like_count",
    "comment_count",
    "subscriber_count",
    "video_count"
]

correlation_matrix = df[correlation_cols].corr()

correlation_matrix
```

```python
# Visualize the correlation matrix

plt.figure(figsize=(8, 6))

plt.imshow(correlation_matrix, interpolation="nearest")
plt.colorbar()

plt.xticks(
    range(len(correlation_cols)),
    correlation_cols,
    rotation=45,
    ha="right"
)

plt.yticks(
    range(len(correlation_cols)),
    correlation_cols
)

plt.title("Correlation Matrix")

for i in range(len(correlation_cols)):
    for j in range(len(correlation_cols)):
        plt.text(
            j,
            i,
            f"{correlation_matrix.iloc[i, j]:.2f}",
            ha="center",
            va="center"
        )

plt.tight_layout()
plt.show()
```

### Engagement Rate Analysis

```python
# Calculate engagement rates

df["like_rate"] = (
    df["like_count"] / df["view_count"]
)

df["comment_rate"] = (
    df["comment_count"] / df["view_count"]
)

df["engagement_rate"] = (
    (df["like_count"] + df["comment_count"])
    / df["view_count"]
)

print("Engagement rate statistics:")
print(
    df[
        ["like_rate", "comment_rate", "engagement_rate"]
    ].describe()
)
```

```python
# Check for invalid engagement rates

rate_columns = [
    "like_rate",
    "comment_rate",
    "engagement_rate"
]

print("Invalid values:")
print(df[rate_columns].isna().sum())

print("\nInfinite values:")
print(
    df[rate_columns]
    .isin([float("inf"), float("-inf")])
    .sum()
)
```

```python
# Visualize engagement rate distribution

plt.figure(figsize=(10, 6))

df["engagement_rate"].plot(
    kind="hist",
    bins=50
)

plt.title("Distribution of Engagement Rate")
plt.xlabel("Engagement Rate")
plt.ylabel("Number of Videos")

plt.tight_layout()
plt.show()
```

```python
# Top videos by engagement rate

top_engagement = (
    df[
        [
            "title",
            "view_count",
            "like_count",
            "comment_count",
            "engagement_rate"
        ]
    ]
    .sort_values(
        "engagement_rate",
        ascending=False
    )
    .head(10)
)

print("Top 10 Videos by Engagement Rate:")
print(top_engagement.to_string(index=False))
```

### Key Findings

- The dataset contains YouTube video and channel-level information collected using multiple search categories and keywords.
- The dataset contains multiple engagement metrics, including views, likes, and comments.
- Engagement metrics show substantial variation across videos.
- View count is strongly affected by a small number of highly viewed videos, resulting in a skewed target distribution.
- Search categories and keywords are not equally represented because data collection is still incomplete.
- Outlier analysis identified potential extreme values in major numerical engagement and channel metrics.
- The correlation analysis provides insight into relationships among views, likes, comments, and other numerical variables.
- Engagement rate provides an additional normalized measure of audience interaction relative to video views.
- The remaining incomplete keyword collection should be completed before performing the final version of EDA.
