**Imports**

```python
import pandas as pd
import requests
import json
```

**Configuration**

```python
from dotenv import load_dotenv
import os

# Load .env file
load_dotenv(override=True)

# Read API key from .env
YOUTUBE_API_KEY = os.getenv("YOUTUBE_API_KEY")

# Optional: Verify it's loaded
print(YOUTUBE_API_KEY[:12])

SEARCH_URL = "https://www.googleapis.com/youtube/v3/search"
```

**Parameters**

```python
params = {
    "part": "snippet",
    "q": "Python Programming",
    "type": "video",
    "maxResults": 5,
    "key": YOUTUBE_API_KEY
}
```

**API Request**

```python
print(response.status_code)
```

```python
response = requests.get(SEARCH_URL, params=params)
print("Key sent:", params["key"][:12])
print("Final URL:", response.request.url)
print(response.text)
```

```python
if response.status_code != 200:
    print("Status Code:", response.status_code)
    print("Response:")
    print(response.text)

response.raise_for_status()
```

**Convert to JSON**

```python
data = response.json()
```

**Inspect JSON**

```python
print(json.dumps(data, indent = 2))
```

**Videos API ENDPOINT**

```python
VIDEOS_URL = "https://www.googleapis.com/youtube/v3/videos"
```

**EXTRACT VIDEO IDS**

```python
video_ids = []
for item in data["items"]:
    video_ids.append(item["id"]["videoId"])

print(video_ids)
```

```python
video_ids = ",".join(video_ids)
print(video_ids)
```

**VIDEOS API PARAMETERS**

```python
video_params = {
    "part": "snippet,statistics,contentDetails",
    "id": video_ids,
    "key": YOUTUBE_API_KEY
}
```

**Call Videos API**

```python
video_response = requests.get(VIDEOS_URL, params=video_params)
print(video_response.status_code)
```

```python
video_data = video_response.json()
```

```python
print(json.dumps(video_data, indent = 2))
```

**Channels API ENDPOINT**

```python
CHANNELS_URL = "https://www.googleapis.com/youtube/v3/channels"
```

```python
channel_ids = []

for item in video_data["items"]:
    channel_ids.append(item["snippet"]["channelId"])

print(channel_ids)
```

```python
channel_ids = list(set(channel_ids))

print(channel_ids)
```

```python
channel_ids = ",".join(channel_ids)

print(channel_ids)
```

```python
channel_params = {
    "part": "snippet,statistics",
    "id": channel_ids,
    "key": YOUTUBE_API_KEY
}
```

```python
channel_response = requests.get(CHANNELS_URL, params=channel_params)
print(channel_response.status_code)
```

```python
channel_data = channel_response.json()
```

```python
print(json.dumps(channel_data, indent = 2))
```
